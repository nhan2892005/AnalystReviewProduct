```python
import os
import sys
import pandas as pd

sys.path.append('./TransferData')

from insert_data_products import get_Product_dataFrame
from insert_data_reviews import get_Review_df, get_Comment_df
```

```python
product_categories_urlKey = pd.read_csv('./CrawlData/CSVFile/URLKeys.csv')
product_categories_urlKey = product_categories_urlKey['URLKeys'].tolist()
```

```python
product = get_Product_dataFrame(product_categories_urlKey)
review = get_Review_df()
comment = get_Comment_df()
```

```python
pro_id = product['product_id'].tolist()
```

```python
import json

def classify_rating(rating):
    if rating > 4:
        return 'tích cực'
    elif rating >= 3:
        return 'trung lập'
    else:
        return 'tiêu cực'

result = []
for id in pro_id[:]:
    if id in review['product_id'].tolist():
        review_of_product = review[review['product_id'] == id]
        
        product_data = {
            'product_id': id,
            'reviews': {
                'tích cực': [],
                'trung lập': [],
                'tiêu cực': []
            }
        }
        
        for _, row in review_of_product.iterrows():
            content = row['content']
            rating = row['rating']
            sentiment = classify_rating(rating)
            if content == '':
                continue
            
            product_data['reviews'][sentiment].append(content)
        
        result.append(product_data)

# Ghi kết quả vào file JSON
with open('product_reviews_classified.json', 'w', encoding='utf-8') as f:
    json.dump(result, f, ensure_ascii=False, indent=4)

print("Dữ liệu đã được ghi vào file 'product_reviews_classified.json'")
```
